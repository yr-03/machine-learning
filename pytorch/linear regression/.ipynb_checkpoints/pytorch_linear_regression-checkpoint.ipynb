# Simple Linear Regression with Pytorch

```python
# importing packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import torch
import torch.nn as nn
print("import success")
```

```python
df = pd.read_csv("customers.csv")
df = df.drop(labels=["Email", "Address", "Avatar"], axis=1)
df.head()
```

```python
df.describe()
```

```python
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
```

```python
plt.scatter(df["Length of Membership"], df["Yearly Amount Spent"])
plt.xlabel("Length of Membership")
plt.ylabel("Yearly Amount Spent")
plt.show()
```

There is quite a strong linear relationship between length of membership and yearly amount spent. Let's try to predict the yearly amount spent from the length of membership.

## Split and Preprocess Data

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
print("import success")
```

```python
X_train_raw, X_test_raw, y_train_raw, y_test_raw = train_test_split(
    df[['Length of Membership']], df['Yearly Amount Spent'], test_size=0.2, random_state=42
)

print(type(X_train_raw))
print(type(X_test_raw))
print(type(y_train_raw))
print(type(y_test_raw))
```

```python
print(f"Length of training data: {len(X_train_raw)}")
print(f"Length of training data targets: {len(y_train_raw)}")
print(f"Length of testing data: {len(X_test_raw)}")
print(f"Length of testing data targets: {len(y_test_raw)}")
```

```python
scaler = StandardScaler()
# sufficient to just scale the inputs
X_train_scaled = scaler.fit_transform(X_train_raw)
X_test_scaled = scaler.transform(X_test_raw)

# convert into tensors to feed into model
X_train = torch.tensor(X_train_scaled, dtype=torch.float32)
X_test = torch.tensor(X_test_scaled, dtype=torch.float32)
y_train = torch.tensor(y_train_raw.values, dtype=torch.float32).view(-1, 1)
y_test = torch.tensor(y_test_raw.values, dtype=torch.float32).view(-1, 1)

print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

## Create Model

```python
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(1, 1)  # 1 input, 1 output

    def forward(self, x):
        return self.linear(x)
```

```python
model = LinearRegressionModel()
criterion = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
```

$$\text{Loss Function, }L = \frac{1}{n}\sum_{i=1}^n(y_{pred_i} - y_{train_i})^2$$

$$\frac{\partial L}{\partial y_{pred}} = \frac{2}{n} \sum_{i=1}^n (y_{pred_i} - y_{train_i})$$

$$ y_{pred} = \hat{\beta_0} + \hat{\beta_1} x$$

$$\frac{\partial L}{\partial \hat{\beta_1}} = 
    \frac{\partial L}{\partial y_{pred}} \frac{\partial y_{pred}}{\partial \hat{\beta_1}}
    = \frac{2}{n} \sum_{i=1}^n (y_{pred_i} - y_{train_i}) \cdot x_i$$

$$\frac{\partial L}{\partial \hat{\beta_0}} = 
    \frac{\partial L}{\partial y_{pred}} \frac{\partial y_{pred}}{\partial \hat{\beta_0}}
    = \frac{2}{n} \sum_{i=1}^n (y_{pred_i} - y_{train_i})$$

$$ \hat{\beta}_{new} = \hat{\beta}_{old} - (lr \cdot \frac{\partial L}{\partial \hat{\beta}}) $$

```python
epochs = 1001

for epoch in range(epochs):
    model.train()  # model enter training mode
    y_pred = model(X_train)  # if it is the first pass, then random weights
    loss = criterion(y_pred, y_train)  # calculate the error from the prediction vs actual label
    optimizer.zero_grad()  # sets the gradient to zero before starting back prop
    loss.backward()  # back prop
    optimizer.step()  # adjusts the weights
    if not epoch % 100:
        print(f'Epoch {epoch}: Loss = {loss.item():.4f}')
```

## Model Evaluation

```python
model.eval()  # switch to evaluation mode

with torch.no_grad():  # disable gradient calculation
    test_pred = model(X_test)  # predict y from x
    test_loss = criterion(test_pred, y_test)  # compare prediction with actual label

print(f"Test Loss: {test_loss.item():.4f}")
```

```python
X_test_np = X_test.detach().numpy() 
y_test_np = y_test.detach().numpy()
preds_np = test_pred.detach().numpy()

plt.scatter(X_test_np, y_test_np, label="Actual Data", color='blue')
plt.plot(X_test_np, preds_np, label='Model Prediction', color='red')
plt.xlabel("Scaled Length of Membership")
plt.ylabel("Yearly Amount Spent")
plt.legend()
plt.show()
```

## Prediction

```python
raw_membership_length = pd.DataFrame([[5.0]], columns = ['Length of Membership'])
# scale and convert into a tensor
scaled_membership_length = torch.tensor(scaler.transform(raw_membership_length), dtype=torch.float32) 

model.eval()
with torch.no_grad():
    prediction = model(scaled_membership_length)
    print(f"Prediction for new membership length: {prediction.item():.2f}")
```

# Multivariate Linear Regression with Pytorch

From the correlation matrix, time on app also has quite a strong correlation with yearly amount spent.

```python
plt.scatter(df['Time on App'], df['Yearly Amount Spent'])
plt.xlabel("Time on App")
plt.ylabel("Yearly Amount Spent")
plt.show()
```

## Split and Preprocess Data

```python
X_train_raw, X_test_raw, y_train_raw, y_test_raw = train_test_split(
    df[['Length of Membership', 'Time on App']], df['Yearly Amount Spent'], test_size=0.2, random_state=42
)

print(type(X_train_raw))
print(type(X_test_raw))
print(type(y_train_raw))
print(type(y_test_raw))
```

```python
print(f"Length of training data: {len(X_train_raw)}")
print(f"Length of training data targets: {len(y_train_raw)}")
print(f"Length of testing data: {len(X_test_raw)}")
print(f"Length of testing data targets: {len(y_test_raw)}")
```

```python
scaler = StandardScaler()
# sufficient to just scale the inputs
X_train_scaled = scaler.fit_transform(X_train_raw)
X_test_scaled = scaler.transform(X_test_raw)

# convert into tensors to feed into model
X_train = torch.tensor(X_train_scaled, dtype=torch.float32)
X_test = torch.tensor(X_test_scaled, dtype=torch.float32)
y_train = torch.tensor(y_train_raw.values, dtype=torch.float32).view(-1, 1)
y_test = torch.tensor(y_test_raw.values, dtype=torch.float32).view(-1, 1)

print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

## Create Model

```python
class MultiLinearRegressionModel(nn.Module):
    def __init__(self, num_inputs):
        super().__init__()
        self.linear = nn.Linear(num_inputs, 1)  # 2 inputs, 1 output

    def forward(self, x):
        return self.linear(x)
```

```python
model = MultiLinearRegressionModel(2)
criterion = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
```

From simple linear regression to multivariate linear regression, the only changes are more variables.

$$ y_{pred_i} = \hat{\beta_0} + \hat{\beta_1} x_1 + \hat{\beta_2} x_2$$

Instead of only $\beta_1$, we can generalise to $\beta_j$.

$$\frac{\partial L}{\partial \hat{\beta_j}} = 
    \frac{\partial L}{\partial y_{pred_i}} \frac{\partial y_{pred_i}}{\partial \hat{\beta_j}}
    = \frac{2}{n} \sum_{i=1}^n (y_{pred_i} - y_{train_i}) \cdot x_j$$

```python
epochs = 1001

for epoch in range(epochs):
    model.train()  # model enter training mode
    y_pred = model(X_train)  # if it is the first pass, then random weights
    loss = criterion(y_pred, y_train)  # calculate the error from the prediction vs actual label
    optimizer.zero_grad()  # sets the gradient to zero before starting back prop
    loss.backward()  # back prop
    optimizer.step()  # adjusts the weights
    if not epoch % 100:
        print(f'Epoch {epoch}: Loss = {loss.item():.4f}')
```

## Model Evaluation


```python
model.eval()  # switch to evaluation mode

with torch.no_grad():  # disable gradient calculation
    test_pred = model(X_test)  # predict y from x
    test_loss = criterion(test_pred, y_test)  # compare prediction with actual label

print(f"Test Loss: {test_loss.item():.4f}")
```

## Prediction

```python
raw_data = pd.DataFrame([[5.0, 11]], columns = ['Length of Membership', 'Time on App'])
# scale and convert into a tensor
scaled_data = torch.tensor(scaler.transform(raw_data), dtype=torch.float32) 

model.eval()
with torch.no_grad():
    prediction = model(scaled_data)
    print(f"Prediction for new data: {prediction.item():.2f}")
```
